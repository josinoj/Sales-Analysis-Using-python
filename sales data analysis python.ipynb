# Sales Data Analysis Using Python

The dataset contains information about product IDs, sales dates, sales representatives, regions, sales amounts, quantities sold, product categories, unit prices, and customer types. Through data cleaning, exploratory data analysis, statistical analysis, and data visualization, this project aims to identify sales patterns, compare regional performance, and understand product category performance.

## Dataset Description

The dataset used in this project contains sales-related information that helps analyze business performance. It includes details about products, sales dates, sales representatives, regions, sales amounts, quantities sold, and product categories.

Dataset Features

Product_ID: Unique identifier for each product.

Sales_Date: Date on which the sale occurred.

Sales_Rep: Name or identifier of the sales representative.

Region: Geographical region where the sale occurred.

Sales_Amount: Total monetary value of the sale.

Quantity_Sold: Number of units sold.

Product_Category: Category to which the product belongs.

This dataset will be used to explore sales patterns, compare regional performance, identify popular product categories, and generate meaningful business insights.

## Bivarate Analysis

### Box Plot of Sales Amount by Product Category


```python
plt.figure(figsize=(9,5))

sns.boxplot(data=df, x='Product_Category',y='Sales_Amount')
plt.title('Sales Amount by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Sales Amount')
plt.xticks(rotation=45)
plt.show()
```

### Analyzing Sales by Product Category

```python
df.groupby('Product_Category')['Sales_Amount'].sum().sort_values(ascending=False)
```

```python
sales_by_category = df.groupby('Product_Category')['Sales_Amount'].sum()
sales_by_category = sales_by_category.sort_values(ascending=False)

sales_by_category.plot(kind = 'bar', figsize=(8,5))

plt.title('Total Sales by Product category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45)
plt.grid(axis='y',alpha=0.3)
plt.show()
```

### Analyzing Sales by Region

```python
sales_by_region= df.groupby('Region')['Sales_Amount'].sum()

sales_by_region = sales_by_region.sort_values(ascending=False)

print(sales_by_region)
```

```python
sales_by_region = df.groupby('Region')['Sales_Amount'].sum()
sales_by_region = sales_by_region.sort_values(ascending=False)

sales_by_region.plot(kind = 'bar', figsize=(8,5))

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=45)
plt.grid(axis='y',alpha=0.3)
plt.show()
```

```python
df.columns
```

```python
top_5_products =(df.groupby('Product_ID')['Sales_Amount'].sum().sort_values(ascending=False).head(5))

print(top_5_products)
```

```python
top_5_products.plot(kind = 'bar', figsize=(8,5))

plt.title('Top 5 Products by Total Sales')
plt.xlabel('Product_ID')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=0)
plt.grid(axis='y',alpha=0.3)
plt.show()
```

### Monthly Sales Trend

```python
monthly_sales = df.groupby('Month')['Sales_Amount'].sum()

monthly_sales.plot(
    kind='line',
    marker='o',
    figsize=(10, 5)
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales Amount')
plt.xticks(range(1, 13))
plt.grid(True, alpha=0.3)
plt.show()
```

### Analyzing Sales by Customer Type

```python
sales_by_customer = df.groupby('Customer_Type')['Sales_Amount'].sum()

sales_by_customer = sales_by_customer.sort_values(
    ascending=False
)

print(sales_by_customer)
```

```python
sales_by_customer.plot(
    kind='bar',
    figsize=(3,5)
)

plt.title('Total Sales by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Total Sales Amount')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.show()
```
